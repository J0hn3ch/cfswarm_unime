**Battery chart:** 
1. Visualize `batteryLevel` and `batteryVoltage`. 
2. Set thresholds 'Low Voltage' and 'Critical Low Voltage' from parameters `pm.lowVoltage` and `pm.criticalLowVoltage`

```python
# Battery chart
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(7, 4))

x = np.linspace(0, 2 * np.pi, 300) # timestamp
y = np.sin(x) * 0.7 + 1.2 # Current
#y2 = y1 # Voltage
#threshold_1 = float( scf.cf.param.get_value(complete_name="pm.lowVoltage") )
#threshold_2 = float( scf.cf.param.get_value(complete_name="pm.criticalLowVoltage") )

ax.plot(x, y, color='#2A9D8F', lw=2)
ax.fill_between(x, 0.5, alpha=0.25, color='orange')
ax.fill_between(x, 0.30, alpha=0.25, color='red')
plt.axhline(y=0.50, color='orange', linestyle='-')
plt.axhline(y=0.30, color='red', linestyle='-')
plt.show()

# Sources: 
# * https://www.pythoncharts.com/matplotlib/matplotlib-gradient-fill-under-line/
# * https://www.geeksforgeeks.org/python/plot-a-horizontal-line-in-matplotlib/
```
