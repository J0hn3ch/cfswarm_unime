```python
import numpy as np
from vispy import app, scene
import jupyter_rfb

# Forza VisPy a usare il backend jupyter_rfb
app.use_app('jupyter_rfb')

# Crea la canvas (la "tela" dove disegnare)
canvas = scene.SceneCanvas(keys='interactive', show=True)
view = canvas.central_widget.add_view()
view.camera = 'panzoom'

# Crea una linea per il sensore
line = scene.visuals.Line(color='cyan', parent=view.scene)

# Inizializza un buffer di dati (es. gli ultimi 1000 campioni)
buffer_size = 1000
x_data = np.linspace(0, 10, buffer_size)
y_data = np.zeros(buffer_size)
pos = np.column_stack((x_data, y_data))
```

```python
def update(ev):
    global pos
    # Qui simuleremo la ricezione di ~16 nuovi campioni 
    # (500Hz / 30 FPS = ~16.6 campioni per frame visivo)
    new_samples = 16
    
    # FAI SCORRERE I DATI VECCHI:
    pos[:-new_samples, 1] = pos[new_samples:, 1]
    
    # INSERISCI I NUOVI DATI (Sostituisci questo con i dati reali del tuo sensore)
    pos[-new_samples:, 1] = np.random.normal(0, 0.1, new_samples)
    
    # Aggiorna la grafica in un colpo solo
    line.set_data(pos=pos)

# Imposta il timer per aggiornare a ~30 FPS
timer = app.Timer(interval=1/30.0, connect=update, start=True)

# Mostra la canvas nella cella
canvas
```
